```python
ap = 1000
pp = 2000
mp = 3000

na = input('사과 갯수 입력')
np = input('배 갯수 입력')
nm = input('메론 갯수 입력')

na = int(na)
np = int(np)
nm = int(nm)

t = (ap * na) * (1-0.1) + pp*np + mp*nm
print('전체 금액', int(t))
```
